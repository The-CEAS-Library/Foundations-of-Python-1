#### Chemistry Solution
Water = 1000
<br>
Isopropyl Alcohol = 7%
<br>
Hydrogen Peroxide = 13%
<br>
Find isoproyl alcohol amount, volume after isopropyl alcohol, hydrogen peroxide amount, total volume, and water percentage

```python
# Create a variable for water, isopropyl alcohol and hydrogen peroxide and assign their values
# Note that percentage is not captured in programming languages. Divide by 100 instead.
water = 1000
alc_percent = 7 / 100
perox_percent = 13 / 100

# Calculate isopropyl alcohol amount by creating a variable for it and assign it to a formula using the variable for water
alc =  water * alc_percent

# Calculate volume of solution after adding isopropyl alcohol
solution1 = water + alc

# Calculate hydrogen peroxide amount
perox = solution1 * perox_percent

# Calculate total amount
total = solution1 + perox

# Calculate water percent rounded to 3 decimal places
water_percent = (water / total) * 100
rounded_water = round(water_percent, 3)

print('The amount of water is', water, 'mL')

# Follow the example above and print the isopropyl alcohol amount
print('The amount of isopropyl alcohol is', alc, 'mL')

# Print the volume after adding isopropyl alcohol
print('The volume after adding isopropyl alcohol is', solution1, 'mL')

# Print the hydrogen peroxide amount
print('The amount of hydrogen peroxide is', perox, 'mL')

# Print the total volume of the solution
print('The total volume is', total, 'mL')

# Print the percentage of water
print('The percentage of water is', rounded_water, 'percent')
```

#### **(Strings, Lists and Dictionaries)**
*   Create a variable with the string value - **'The quick brown fox jumps over a lazy dog'** <br>
    * Print the length of this string variable
    * **'Split'** the string using **'  '** (space character) and assign this operation to a new variable. Print the length of the new variable
    * Find the middle element of the sequence in new variable and return its value
    * **'append'** a new string value, 'number 5' (5 is an integer. Concatenate the values, 'number' and 5, while appending)
    * **'Join'** the split strings (in the new variable) with a space character and print this operation (**Hint: ' '.join(???)**)



```python
text = 'The quick brown fox jumps over a lazy dog'

#Length of the 'text' variable
print('Length of the variable "text":', len(text))

#Splitting the string
new = text.split(' ')

#Length of the 'new' variable
print('Length of the variable "new":', len(new))

#Middle element of the 'new' variable
mid_obj = len(new)//2
print('Middle Element in the new variable: ', new[mid_obj])

#Adding a string to the 'new' variable
new.append('number ' + str(5))
print(new)

#Joining the the individual strings into one string
print(' '.join(new))
```
